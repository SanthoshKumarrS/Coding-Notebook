# Matplotlib Fundamentals

## 1. Overview & Setup
* **Definition:** Matplotlib is a Python library for data visualization, ranging from basic line graphs to complex data plots.
* **Standard Import:** `import matplotlib.pyplot as plt`
* **Basic Plotting:** `plt.plot(x, y)` creates a line graph from 1D arrays or lists.
* **Displaying Graphs:** Use `plt.show()` to display the graph cleanly and hide unnecessary memory address text in tools like Jupyter Notebook.

---

## 2. Labels, Titles & Legends
* **Titles and Axes:** Context can be added using `plt.title()`, `plt.xlabel()`, and `plt.ylabel()`.
* **Font Customization:** Pass a `fontdict` parameter to titles and labels to modify font name, size, and weight (e.g., bold).
* **Tick Marks:** `plt.xticks()` and `plt.yticks()` specify exact values displayed on axes and assist with overall scaling.
* **Legends:** Include a `label` argument inside the plot function and execute `plt.legend()` to distinguish multiple lines.

---

## 3. Styling & Shorthand Notation
* **Explicit Parameters:** Customize line appearance using keyword arguments:
  * `color` (names or hex codes)
  * `linewidth`
  * `marker` (e.g., dots or circles)
  * `linestyle`
* **Shorthand Syntax:** Combine styling using the format `'[color][marker][line style]'` (e.g., `'r.--'` for a red, dotted, dashed line).
* **Multiple Lines:** Call `plt.plot()` multiple times within the same cell to draw multiple lines on one graph.

---

## 4. Common Chart Types
* **Bar Charts:** Created using `plt.bar(labels, values)`. Individual bars accept patterns (hatches) like stripes or stars via `.set_hatch()`.
* **Histograms:** Display data distribution across grouped "bins" using `plt.hist(data, bins=[list])`.
* **Pie Charts:** Created using `plt.pie(values, labels=labels)`. Supports percentage formatting (`autopct`) and slice separation (`explode`).
* **Box and Whisker Plots:** Compare datasets using `plt.boxplot()` to display median, minimum, maximum, and interquartile range (middle 50%).

---

## 5. Figure Management & Saving
* **Resizing:** Call `plt.figure(figsize=(width, height), dpi=pixels_per_inch)` at the beginning of the code block to configure dimension and resolution.
* **Resolution (DPI):** Higher DPI values create sharper, higher-resolution images for printing or zooming.
* **Exporting:** Save visual output as image files using `plt.savefig('filename.png')`.

---

## 6. Integration with NumPy & Pandas
* **NumPy:** Generate continuous mathematical datasets (e.g., `np.arange()`) to plot functions like curves or sine waves.
* **Pandas:** Load CSV files using `pd.read_csv()` and plot DataFrame columns directly (e.g., `plt.plot(df.Year, df.Price)`).
* **Data Preparation:** Clean data using Pandas prior to plotting (e.g., filtering columns, stripping string units like `'lbs'`, and casting data types).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = [0, 1, 2, 3,4]
y = [2, 4, 6,8,10]

# resize the graph

plt.figure(figsize=(7,3.5),dpi=150)

#plt.plot(x,y, label = '2x', color = "purple",linewidth = 2.5,marker = '.',markersize = 10,linestyle = '--',markeredgecolor = 'red')
plt.plot(x,y, 'b^--',label = '2x')
plt.title('First Title', fontdict = {'fontname': 'sans-serif', 'fontsize': 15})
plt.xlabel('X Axis')
plt.ylabel('Y Axis')

#Plot part of the graph as line
x2 = np.arange(0,4.5,0.5)
plt.plot(x2[:7], x2[:7]**2,'r',label = 'X^2')
#Pot reminder as dots
plt.plot(x2[4:], x2[4:]**2,'r--',label = 'X^2')

plt.xticks([0,1,2,3,4])
plt.yticks([2,4,6,8,10])

plt.legend()

#Saves the chart in the working directory
plt.savefig('mygraph.png',dpi = 300)

plt.show()
```

```python
labels = ['A','B','C']
values = [1,4,2]

bars = plt.bar(labels,values)

bars[0].set_hatch('/')
bars[1].set_hatch('o')
bars[2].set_hatch('*')

plt.figure(figsize=(6,4))

plt.show()
```

```python
gas = pd.read_csv('gas_prices.csv')
gas.head()
```

```python
plt.title('Gas Price over time in USD')

plt.plot(gas.Year,gas.USA,'b.-',label = 'USA')
plt.plot(gas.Year,gas.Canada,'r.-',label = 'Canada')
plt.plot(gas.Year,gas['South Korea'],'g.-',label = 'South Korea')
plt.plot(gas.Year,gas.Australia,'y.-',label = 'Australia')

plt.xticks(gas.Year[::4])
plt.xlabel('Year')
plt.ylabel('Price in USD')

plt.legend(bbox_to_anchor=(1, 1))
plt.show()
```

```python
plt.title('Gas Price over time in USD', fontdict={'fontweight':'bold', 'fontsize':20})
plt.grid()
for country in gas:
    if country != 'Year':
        plt.plot(gas.Year, gas[country],marker = '.',label = country)

plt.xticks(gas.Year[::4].tolist()+ [2011])
plt.xlabel('Year')
plt.ylabel('Price in USD')

plt.legend(bbox_to_anchor = (1,1))

plt.savefig('Gas_price-fig.png',dpi = 300)
plt.show()
```

### Load Fifa Data

```python
fifa = pd.read_csv('fifa_data.csv')
fifa.head()
```

#### `Histograms`

```python
bins = [40,50,60,70,80,90,100]
plt.hist(fifa.Overall, bins = bins,color='red',edgecolor='black')
plt.title('FIFA Players Overall Rating Distribution')
plt.xticks(bins)

plt.xlabel('skill level')
plt.ylabel('Number of players')

plt.show()
```

#### `Pie-Charts`

```python
one = 10
two = 20
three = 30
four = 40

num = [one,two,three,four]
labels = ['one','two','three','four']

plt.pie(num,labels=labels)
plt.show()
```

```python
foot = fifa['Preferred Foot'].value_counts()

colors = ['#abcdef', '#aabbcc']

plt.pie(foot, labels=foot.index, colors=colors,autopct= '%.2f%%')
plt.title('Fifa Players Preferred Foot Distribution')

plt.show()
```

```python
fifa.Weight = [int(x.strip('lbs')) if type(x)== str 
               else x for x in fifa.Weight]

fifa.Weight[0]
```

```python
plt.style.use('ggplot')

light = fifa.loc[fifa.Weight < 125].shape[0]
light_medium = fifa.loc[(fifa.Weight >= 125) & (fifa.Weight < 150)].shape[0]
medium = fifa.loc[(fifa.Weight >= 150) & (fifa.Weight < 175)].shape[0]
medium_heavy = fifa.loc[(fifa.Weight >= 175) & (fifa.Weight < 200)].shape[0]
heavy = fifa.loc[fifa.Weight >= 200].shape[0]

weights = [light, light_medium, medium, medium_heavy, heavy]
labels = ['Under 125 lbs', '125-150 lbs','150-175 lbs','175-200 lbs', 'over 200 lbs']
explode = (.35,.2,0,.1,.35)

plt.pie(weights,labels=labels, autopct= '%.2f%%',explode = explode)
```

#### `Box-Charts`

```python
barcelona = fifa.loc[fifa.Club == 'FC Barcelona']['Overall']
madrid = fifa.loc[fifa.Club == 'Real Madrid']['Overall']
labels = ['Barcelona','Madrid']
plt.title('Rating for FC Barcelona vs Real Madrid')

plt.boxplot([barcelona,madrid], tick_labels= labels,patch_artist=True,medianprops={'color':'black'})
plt.show()
```
