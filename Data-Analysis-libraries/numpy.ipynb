# 📘 NumPy Fundamentals

---

## 🔹 What is NumPy?

- Core library for **Scientific Computing** in Python.
- Foundation for **Data Science**, **Machine Learning**, and **Deep Learning**.
- Provides a fast, efficient **N-dimensional array (`ndarray`)**.
- Over **100× faster** than Python lists for numerical operations.
- Standard import:

```python
import numpy as np
```

---

# 🧩 The `ndarray`

The **ndarray** is NumPy's central object.

### Important Attributes

| Attribute | Description |
|-----------|-------------|
| `shape` | Dimensions of the array |
| `dtype` | Data type of elements |
| `size` | Total number of elements |
| `itemsize` | Size of each element (bytes) |
| `ndim` | Number of dimensions |

Example:

```python
a.shape
a.dtype
a.size
a.itemsize
a.ndim
```

# ⚡ NumPy vs Python Lists

| NumPy Arrays | Python Lists |
|--------------|--------------|
| Faster (C implementation) | Slower |
| Element-wise operations | List repetition |
| Contiguous memory | Non-contiguous memory |
| No `.append()` | Supports `.append()` |

Example:

```python
np.array([1,2,3]) * 2
# [2 4 6]

[1,2,3] * 2
# [1,2,3,1,2,3]
```

# 📋 Copy vs Reference

❌ Reference Copy

```python
b = a
```

Both variables point to the **same array**.

---

✅ Independent Copy

```python
b = a.copy()
```

Changes to one array **do not affect** the other.

---

# 💡 Quick Tips

- ✅ Import as:

```python
import numpy as np
```

- ✅ Use `.copy()` for independent arrays.
- ✅ Prefer `@` for matrix multiplication.
- ✅ Use `reshape()` instead of creating new arrays.
- ✅ Use `axis=0` → Columns.
- ✅ Use `axis=1` → Rows.
- ✅ Broadcasting makes NumPy concise and efficient.

# 🚀 Most Frequently Used NumPy Functions

```python
np.array()          # Create a NumPy array from a Python list or tuple

np.zeros()          # Create an array filled with zeros

np.ones()           # Create an array filled with ones

np.full()           # Create an array filled with a specified value

np.eye()            # Create an identity matrix

np.arange()         # Generate evenly spaced values within a given range

np.linspace()       # Generate evenly spaced values between two endpoints

np.random.random()  # Generate random numbers between 0 and 1

np.random.randn()   # Generate random numbers from a normal (Gaussian) distribution

np.reshape()        # Change the shape of an array without changing its data

np.concatenate()    # Join two or more arrays along an existing axis

np.hstack()         # Stack arrays horizontally (column-wise)

np.vstack()         # Stack arrays vertically (row-wise)

np.dot()            # Compute the dot product of two arrays

@                   # Matrix multiplication operator

np.linalg.inv()     # Compute the inverse of a square matrix

np.linalg.det()     # Compute the determinant of a matrix

np.linalg.eig()     # Compute eigenvalues and eigenvectors

np.linalg.solve()   # Solve a system of linear equations (Ax = b)

np.sum()            # Compute the sum of array elements

np.mean()           # Compute the average (mean)

np.std()            # Compute the standard deviation

np.var()            # Compute the variance

np.max()            # Return the maximum value

np.min()            # Return the minimum value

np.argmax()         # Return the index of the maximum value

np.argmin()         # Return the index of the minimum value

np.sort()           # Sort an array

np.unique()         # Return unique elements in an array

np.allclose()       # Check if two arrays are approximately equal

np.where()          # Return indices or values based on a condition

np.copy()           # Create an independent copy of an array

np.loadtxt()        # Load numerical data from a text or CSV file

np.genfromtxt()     # Load data from a file while handling missing values
```

````python
import numpy as np

print(f'Numpy Version : {np.__version__}')
````

````python
a = np.array([1, 2, 3, 4, 5])
print(a)
print(a.shape)
print(a.dtype)
print(a.ndim)
print(a.size)
print(a.itemsize) # returns in bytes
````

````python
print(a[0])  # Accessing the first element

a[0] = 10  # Modifying the first element

print(a)  # Printing the modified array
````

````python
#Dot Product

b = np.array([6, 7, 8, 6, 1])

dot = np.dot(a, b)
print(f'Dot Product : {dot}')

sum1 = a * b
dot = np.sum(sum1)
print(f'Dot Product : {dot}')

dot2 = a @ b
print(f'Dot Product : {dot2}')
````

````python
#Multidimensional (nd) arrays 

c = np.array([[1, 2], [3, 4]])
print(c)
print(c.shape)
print(c[0,1])
print(np.linalg.inv(c))
````

````python
#Where  
print(f"c = {c[c > 2]}")

d = np.where(c > 2, c, 0)
print(f"d = {d}")
````

````python
#Fancy Indexing
a = np.array([1, 2, 3, 4, 5])
indices = [0, 2, 4]
fancy_indexed_array = a[indices]
print(f"Fancy Indexed Array: {fancy_indexed_array}")
````

````python
e = np.arange(1,9)
print(f"e = {e}")

print(e.shape)
b = e.reshape(2,4)
print(b)
````

````python
#Concatenate

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6]])
c = np.concatenate((a, b), axis=0)
print(f"Concatenated Array: \n{c}")
````

````python
d = np.array([[1,2,5,8,9,10], [10,20,30,4,50,62]])
print(d)
print(d.sum(axis=1))  # Sum along columns
print(d.mean(axis=1).round(2))  # Mean along columns
````

````python
x = np.array([1,2])
print(x)
print(x.dtype)
````

````python
#Generating arrays
a = np.zeros((2, 3))
print(f"Array of zeros: \n{a}")

b = np.ones((3, 2))
print(f"Array of ones: \n{b}")

c = np.full((2, 2), 7)
print(f"Array of sevens: \n{c}")

d = np.eye(3)
print(f"Identity Matrix: \n{d}")

e = np.arange(10)
print(f'arrange numbers = {e}')

f = np.linspace(0, 1, 5)
print(f'linspace numbers = {f}')

g = np.random.rand(2, 3)
print(f'Random numbers = \n{g}')

h = np.random.randint(0, 10, size=(2, 3))
print(f'Random integers = \n{h}')

i = np.random.choice([1, 2, 3, 4, 5], size=(2, 3), replace=True)
print(f'Random choice = \n{i}')
````

````python
#Linear Algebra

a = np.array([[1, 2], [3, 4]])
eigenvalues, eigenvectors = np.linalg.eig(a)

print(f'eigen values = \n{eigenvalues}') 
print(f'eigen vectors = \n{eigenvectors}')
````

````python
a = np.array([1,2,3])
b = np.array([1,2,3])

print(np.allclose(a, b))  # Check if all elements are close to each other
````

````python
#Solving linear equations

A = np.array([[1,1], [1.5,4.0]])
B = np.array([2200,5050])

x = np.linalg.inv(A).dot(B)
print(x)

y = np.linalg.solve(A,B)
print(y)
````
