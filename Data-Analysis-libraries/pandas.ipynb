# 🐼 Pandas Essentials & Quick Reference

---

## 1. 🚀 Basics & Importing

```python
import pandas as pd
```

### 📥 Reading Files
| File Type | Function | Notes |
| :--- | :--- | :--- |
| **CSV** | `pd.read_csv('path.csv')` | Standard delimited tabular data |
| **Excel** | `pd.read_excel('path.xlsx')` | Supports `sheet_name` parameter |
| **JSON / Text** | `pd.read_json()` / `pd.read_table()` | Structured JSON or custom delimited text |

<div class="alert alert-block alert-warning">
  <b>💡 Pro-Tip (Raw Strings):</b> Always prefix Windows paths with <code>r</code> (e.g., <code>r'C:\folder\file.csv'</code>) to prevent backslash <code>\</code> Unicode escape errors.
</div>

> **DataFrames:** Reading data stores it as a **DataFrame**—a 2D labeled data structure with rows and columns.

---

## 2. 🔍 Data Exploration & Inspection

* **Quick Preview:** `df.head(n)` *(first n rows)* | `df.tail(n)` *(last n rows)*

| Attribute / Method | Description |
| :--- | :--- |
| `df.shape` | Returns dimensions as `(rows, columns)` |
| `df.info()` | Summary of data types, memory usage, and non-null counts |
| `df.describe()` | Summary statistics (mean, min, max, percentiles) for numeric columns |
| `df.isnull().sum()` | Counts total missing (`NaN`) values per column |

---

## 3. 🎯 Filtering & Selecting Data

* **Single Column:** `df['ColumnName']`
* **Conditionals:** `df[df['rank'] < 10]`
* **List Matching:** `df[df['category'].isin(['A', 'B'])]`
* **Substring Search:** `df[df['name'].str.contains('text')]`
* **Sorting Data:** `df.sort_values(by='col', ascending=False)`

---

## 4. 📌 Indexing: `.loc` vs `.iloc`

| Method | Indexing Strategy | Example |
| :--- | :--- | :--- |
| **`.loc`** | **Label-based** (uses index names/labels) | `df.loc[0, 'ColumnName']` |
| **`.iloc`** | **Integer position-based** (uses numerical position) | `df.iloc[0, 2]` *(1st row, 3rd column)* |

<details>
<summary><b>🔧 Customizing the Index</b></summary>

```python
df.set_index('ColumnName')  # Set a column as the index
df.reset_index()            # Revert index back to default numeric sequence
```
</details>

---

## 5. 🧹 Data Cleaning

```python
# 1. Remove duplicate rows
df.drop_duplicates()

# 2. Remove unwanted columns
df.drop(columns=['UnwantedColumn'])

# 3. Handle null values
df.fillna('DefaultValue')  # Replace NaNs with a value
df.dropna()                # Drop rows containing any missing values

# 4. String cleaning
df['Col'].str.strip()              # Trim leading/trailing whitespace
df['Col'].str.replace('old', 'new') # Swap specific string values
```

---

## 6. 📊 Grouping & Aggregating

Group data by categorical values using `df.groupby('Column')`:

* **Standard Functions:** `.mean()`, `.sum()`, `.count()`, `.min()`, `.max()`
* **Multi-Calculations (`.agg()`):**

```python
# Apply different functions across multiple columns simultaneously
df.groupby('Category').agg({
    'Sales': 'sum',
    'Age': 'mean',
    'ID': 'count'
})
```

---

## 7. 🔗 Combining DataFrames

### 🔀 Merging (`.merge()`)
Combines DataFrames SQL-style based on shared keys:
```python
df1.merge(df2, on='ID', how='inner')  # Options: 'inner', 'outer', 'left', 'right'
```

### 📚 Concatenating (`.concat()`)
Stacks DataFrames vertically (rows) or horizontally (columns):
```python
pd.concat([df1, df2], axis=0) # Vertical stack
pd.concat([df1, df2], axis=1) # Horizontal join
```

---

## 📈 8. Data Visualization

Generate quick plots directly from your DataFrame using `df.plot()`:

```python
df.plot(

    y='y_column', 
    title='Chart Title',
    xlabel='X-Axis Label',
    ylabel='Y-Axis Label',

```

````python
df1.merge(df2,how = 'right')
````

````python
df1.merge(df2,how = 'cross') # Cross join, returns the cartesian product of both dataframes
````

## Data Cleaning in Pandas

````python
df = pd.read_excel('Customer Call List.xlsx')
df.head()
````

````python
df = df.drop_duplicates()
````

````python
df = df.drop(columns = 'Not_Useful_Column')
df.head()
````

````python
#df['Last_Name'] = df['Last_Name'].str.lstrip("...")
#df['Last_Name'] = df['Last_Name'].str.lstrip("/")
#df['Last_Name'] = df['Last_Name'].str.lstrip("_")
df['Last_Name'] = df['Last_Name'].str.strip("123./_")
df
````

````python
#df['Phone_Number'] =df['Phone_Number'].str.replace('[^a-zA-Z0-9]','',regex=True)

#df['Phone_Number'] = df['Phone_Number'].apply(lambda x: str(x))
#df['Phone_Number'] = df['Phone_Number'].apply(lambda x: x[0:3]+ "-" + x[3:6] + "-" + x[6:10])
df['Phone_Number'] = df['Phone_Number'].str.replace('nan--Na--','')
df['Phone_Number'] = df['Phone_Number'].str.replace('Na--','')
df
````

````python
df[['Street_Address','State','Zip_code']]= df['Address'].str.split(',',n = 2, expand = True)
df
````

````python
df['Paying Customer'] = df['Paying Customer'].str.replace('Yes','Y').replace('No','N')
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('Yes','Y').replace('No','N')
df
````

````python
df = df.replace('N/a','')
df = df.fillna('')
df
````

````python
for x in df.index:
    if df.loc[x,'Do_Not_Contact'] == 'Y':
        df.drop(x, inplace=True)

df
````

````python
for x in df.index:
    if df.loc[x,'Phone_Number'] == '':
        df.drop(x, inplace=True)

df
````

````python
df = df.reset_index(drop = True)
df.head()
````
