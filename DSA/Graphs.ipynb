# 🌐 Graph Fundamentals

## 📌 What is a Graph?

A **Graph** is the most general data structure consisting of:

- **Vertices (Nodes)** → Store data
- **Edges** → Connections between nodes

Unlike Trees, Graphs can have **multiple paths**, **cycles**, and **disconnected components**.

Example:

```text
      A
     / \
    B---C
     \ /
      D
```

Here:

- Vertices = A, B, C, D
- Edges = (A-B), (A-C), (B-C), (B-D), (C-D)

---

# 🧱 Components of a Graph

## 📍 Vertices (Nodes)

Vertices store the actual information.

Example:

```text
A

B

C

D
```

---

## 🔗 Edges

Edges represent connections between two vertices.

Example:

```
A ------ B
```

Edge:

```
(A, B)
```

---

# 🌳 Graph Hierarchy

Many common data structures are actually **special types of Graphs**.

```
Graph

├── Tree

└── Linked List
```

- Every **Tree** is a Graph.
- Every **Linked List** is a Graph.
- But not every Graph is a Tree or Linked List.

---

# 🔗 Connected Graph

A graph is **Connected** if every node can be reached from every other node.

Example:

```text
      A
     / \
    B   C
     \ /
      D
```

Every node is reachable.

---

# ❌ Disconnected Graph

A graph is **Disconnected** if at least one node cannot be reached.

Example:

```text
A ----- B


C ----- D
```

There are two separate components.

---

# 🔄 Cyclic Graph

A graph is **Cyclic** if it contains a path that eventually returns to the starting node.

Example:

```text
      A
     / \
    B---C
```

Cycle:

```
A

↓

B

↓

C

↓

A
```

---

# 🌲 Acyclic Graph

A graph without any cycles.

Example:

```text
      A
     / \
    B   C
        |
        D
```

No path returns to its starting node.

---

# ➡️ Directed Graph (Digraph)

Edges have a direction.

Example:

```text
A ───► B ───► C
```

You can travel:

```
A → B
```

but not necessarily:

```
B → A
```

unless another edge exists.

---

# ↔️ Undirected Graph

Edges have **no direction**.

Example:

```text
A ───── B
```

Means:

```
A ↔ B
```

Movement is possible in both directions.

---

# 📦 Graph Representations

Graphs can be stored in several ways.

---

# 1️⃣ Edge List

Stores every connection as a pair of nodes.

Example:

```text
A ----- B

|

C
```

Edge List:

```python
[
    ("A", "B"),
    ("A", "C")
]
```

Simple but inefficient for searching neighbors.

---

# 2️⃣ Adjacency Matrix

Uses a **2D matrix**.

Rows represent:

```
From Node
```

Columns represent:

```
To Node
```

Example:

```text
      A B C

A     0 1 1

B     1 0 0

C     1 0 0
```

If:

```
Matrix[A][B] = 1
```

then:

```
A is connected to B
```

---

## ⚠️ Limitation

Sparse graphs contain many zeros.

Example:

```
1000 Nodes

↓

Only 10 edges
```

The matrix still stores:

```
1000 × 1000
```

entries.

Memory usage becomes inefficient.

---

# 3️⃣ Adjacency List ⭐ (Most Common)

Stores neighbors for each node.

Python Example:

```python
graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A"],
    "D": ["B"]
}
```

Visualization:

```text
A → B, C

B → A, D

C → A

D → B
```

Advantages:

- Memory efficient
- Fast neighbor lookup
- Most common representation in interviews

---

# 4️⃣ Class-Based Graph

Each node stores references to its neighbors.

```python
class Node:

    def __init__(self, value):

        self.value = value
        self.neighbors = []
```

Example:

```text
A

↓

[B,C]
```

Useful in object-oriented implementations.

---

# 🔍 Graph Traversals

Graph traversal means visiting every node.

The two major traversal algorithms are:

- Depth First Search (DFS)
- Breadth First Search (BFS)

---

# ⚠️ The Seen (Visited) Set

Graphs can contain cycles.

Without tracking visited nodes, traversal may never terminate.

Example:

```text
A

↓

B

↓

C

↓

A
```

Infinite loop:

```
A

↓

B

↓

C

↓

A

↓

B

↓

C

...
```

Solution:

Maintain a **Seen (Visited) Set**.

```python
visited = set()
```

Before visiting a node:

```python
if node in visited:
    return
```

---

# 🌳 Depth First Search (DFS)

DFS explores one branch **as deep as possible** before backtracking.

Example:

```text
      A
     / \
    B   C
   /
  D
```

Traversal:

```
A

↓

B

↓

D

↓

C
```

---

## Recursive DFS

Uses recursion.

```python
def dfs(node):

    if node in visited:
        return

    visited.add(node)

    for neighbor in graph[node]:
        dfs(neighbor)
```

Uses the **Call Stack**.

---

## Iterative DFS

Uses an explicit **Stack**.

```python
stack = [start]
```

Process:

```
Pop

↓

Visit

↓

Push Neighbors
```

---

# 🌊 Breadth First Search (BFS)

BFS visits nodes **level by level**.

Uses a **Queue**.

Example:

```text
      A
     / \
    B   C
   / \
  D   E
```

Traversal:

```
A

↓

B

↓

C

↓

D

↓

E
```

---

## Python Implementation

```python
from collections import deque

queue = deque([start])
```

Process:

```
Remove Front

↓

Visit

↓

Add Neighbors
```

---

# ⏱️ Time Complexity

Both DFS and BFS visit:

- Every Vertex
- Every Edge

Exactly once.

```
Time Complexity

O(V + E)
```

Where:

- **V** = Number of Vertices
- **E** = Number of Edges

---

# 💾 Space Complexity

Need memory for:

- Adjacency List
- Seen Set
- Stack (DFS)
- Queue (BFS)

Overall:

```
O(V + E)
```

---

# 🌳 Trees are Special Graphs

A **Tree** is simply a Graph that is:

- ✅ Connected
- ✅ Acyclic (No cycles)

Example:

```text
        A
       / \
      B   C
         /
        D
```

Properties:

- Every node is reachable.
- No cycles exist.

---

# 📏 Edge Rule for Trees

If a Tree contains:

```
N Nodes
```

then it will always contain:

```
N - 1 Edges
```

Example:

```
Nodes = 5

↓

Edges = 4
```

Example:

```text
        A
       / \
      B   C
     /
    D
   /
  E
```

Nodes:

```
5
```

Edges:

```
4
```

---

# 📊 Graph Storage Comparison

| Representation | Space | Advantages | Disadvantages |
|---------------|:-----:|------------|---------------|
| Edge List | O(E) | Very simple | Slow neighbor lookup |
| Adjacency Matrix | O(V²) | Fast edge lookup | Wastes memory for sparse graphs |
| Adjacency List | O(V + E) | Memory efficient, fast traversal | Slightly slower edge lookup than a matrix |
| Class-Based | O(V + E) | Object-oriented, flexible | More complex implementation |

---

# 📊 DFS vs BFS

| Feature | DFS | BFS |
|----------|:---:|:---:|
| Strategy | Go Deep First | Visit Level by Level |
| Data Structure | Stack / Recursion | Queue |
| Time Complexity | O(V + E) | O(V + E) |
| Space Complexity | O(V + E) | O(V + E) |
| Common Uses | Backtracking, Cycle Detection, Topological Sort | Shortest Path (Unweighted), Level Order Traversal |

# 📝 Key Takeaways

- A **Graph** consists of **Vertices (Nodes)** connected by **Edges**.
- **Trees** and **Linked Lists** are special types of Graphs.
- Graphs can be:
  - **Connected** or **Disconnected**
  - **Cyclic** or **Acyclic**
  - **Directed** or **Undirected**
- Common graph representations:
  - **Edge List**
  - **Adjacency Matrix**
  - **Adjacency List** (most common)
  - **Class-Based Graph**
- Use a **Seen (Visited) Set** during traversal to prevent infinite loops caused by cycles.
- **Depth First Search (DFS)** explores one path as deep as possible using a **Stack** or **Recursion**.
- **Breadth First Search (BFS)** explores level by level using a **Queue**.
- Both DFS and BFS have:
  - **Time Complexity:** **O(V + E)**
  - **Space Complexity:** **O(V + E)**
- A **Tree** is a Graph that is **Connected** and **Acyclic**.
- A Tree with **N nodes** always has exactly **N − 1 edges**.

````python
# Array of Edges (Directed) [Start, End]
n = 8
A = [[0, 1], [1, 2], [0, 3], [3, 4], [3, 6], [3, 7], [4, 2], [4, 5], [5, 2]]

A
````

````python
# Convert Array of Edges -> Adjacency Matrix

M = []

for i in range(n):
    M.append([0] * n)

for u,v in A:
    M[u][v] = 1

M
````

````python
# Convert Array of Edges -> Adjacency List
from collections import defaultdict

D = defaultdict(list)

for u,v in A:
    D[u].append(v)

print(D[3])
````

````python
# DFS with Recursion - O(V + E) where V is the number of nodes and E is the number of edges

def dfs_recursive(node):
    print(node)

    for neighbour_node in D[node]:
        if neighbour_node not in seen:
            seen.add(neighbour_node)
            dfs_recursive(neighbour_node)


source = 0
seen = set()
seen.add(source)
dfs_recursive(source)

````

````python
# Iterative DFS with Stack - O(V + E)
source = 0

seen = set()
seen.add(source)
stack = [source]

while stack:
    node = stack.pop()
    print(node)
    for neighbour_node in D[node]:
        if neighbour_node not in seen:
            seen.add(neighbour_node)
            stack.append(neighbour_node)
````

````python
# BFS (Queue) - O(V + E)

source = 0

from collections import deque

seen = set()
seen.add(source)
q = deque()
q.append(source)

while q:
    node = q.popleft()
    print(node)
    for neighbour_node in D[node]:
        if neighbour_node not in seen:
            seen.add(neighbour_node)
            q.append(neighbour_node)
````
